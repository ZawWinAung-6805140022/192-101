## Problem 1 — Elevator Direction

A building has floors numbered **1 to 30**. Read the **current floor** the elevator is on and the
**target floor** a passenger wants. Decide what the elevator should do and how far it must travel.

**Your program must**
- read `current` and `target` as whole numbers
- if either floor is below 1 **or** above 30, print `Invalid floor` and stop deciding
- otherwise decide:
  - target higher → `Going up N floors`
  - target lower → `Going down N floors`
  - same floor → `You are already on this floor`
- `N` is how many floors it travels (the difference between the two floors)

**Sample runs**
```
Current floor: 3        Current floor: 20       Current floor: 7        Current floor: 35
Target floor: 12        Target floor: 5         Target floor: 7         Target floor: 4
Going up 9 floors       Going down 15 floors    You are already         Invalid floor
                                                on this floor
```

*Hint:* check the range first with `or`. For the distance, subtract the smaller floor from the
larger so `N` is never negative — or compute it inside each branch.

````python
# Write your program here
current_floor=int(input("Current floor"))
target_floor=int(input("Target floor"))
if current_floor <1 or current_floor >30:
    print("Invalid floor")
elif target_floor > current_floor :
    print(f"going up {target_floor-current_floor} floors")
elif target_floor < current_floor :
    print(f"going down {current_floor-target_floor} floors")
else :
    print("you are already on this floor")
````

## Problem 2 — Cinema Ticket Price

A cinema sets ticket prices by age, then adds a weekend surcharge. The **base** price is **120 baht**.
A **child** (age under 12) pays **half** the base. A **senior** (age 60 or over) pays **80 baht**.
Everyone else pays the full 120. On a **weekend**, add **30 baht** to whatever the price is.

**Your program must**
- read the passenger's `age` (whole number)
- read whether it is a weekend as `y` or `n`
- decide the age-based price using `if` / `elif` / `else`
- if it is a weekend, add 30 to that price
- print `Ticket price: X baht`

**Sample runs**
```
Age: 10                 Age: 70                 Age: 30
Weekend (y/n): n        Weekend (y/n): y        Weekend (y/n): n
Ticket price: 60 baht   Ticket price: 110 baht  Ticket price: 120 baht
```

*Hint:* work out the age price first and store it in a variable. Then add the surcharge in a
**separate** `if` that checks the weekend answer — don't mix the two decisions together.

````python
# Write your program here
age=int(input("Age"))
weekend=str(input("weekend y or n:"))
baseprice=120
if age <12 and weekend=="y":
  print(60+30)
  print("Ticket price:60+30=90baht")
elif age <12 and weekend=="n":
  print("Ticket price:60baht")
elif age >=60 and weekend=="y":
  print("Ticket price:80+30=110baht")
elif age >=60 and weekend=="n":
  print("Ticket price:80baht")
else :
  print(f"Ticket price:{baseprice} baht")
````

## Problem 3 — ATM Withdrawal

An ATM only dispenses cash in **multiples of 100 baht**. It refuses a withdrawal if the amount is
**not** a multiple of 100, if it is **more than the balance**, or if it is **more than the 20000 daily limit**.
Read the `balance` and the requested `amount`, then respond.

**Your program must**
- if the amount is not a multiple of 100 (`amount % 100 != 0`) → `Amount must be a multiple of 100`
- else if the amount is more than the balance → `Insufficient funds`
- else if the amount is more than 20000 → `Over daily limit`
- otherwise print `Dispensing X baht. New balance: Y`
- check the reasons **in this order** and print only one message (use `elif`)

**Sample runs**
```
Balance: 5000           Balance: 1000           Balance: 50000          Balance: 5000
Amount: 250             Amount: 2000            Amount: 25000           Amount: 3000
Amount must be a        Insufficient funds      Over daily limit        Dispensing 3000 baht.
multiple of 100                                                         New balance: 2000
```

*Hint:* the `%` operator gives the remainder; a multiple of 100 leaves remainder 0. Because you use
`elif`, the first matching reason wins — so the order of your checks is part of the answer.

````python
# Write your program here
balance=int(input("Balance: "))
amount=int(input("withdraw amount"))
if amount % 100 != 0:
  print ("amount must be multiple of 100")
elif amount > balance:
  print ("insufficient funds")
elif amount > 20000:
  print ("Overr daily limit")
else:
  print(f"Dispensing {amount} baht")
  newbalance=balance-amount
  print(f"New balance {newbalance} baht")

````

## Problem 4 — Electricity Bill

Monthly electricity is billed in **tiers** by the number of units used: the **first 150 units** cost
**3 baht each**; the **next 250 units** (units 151–400) cost **4 baht each**; every unit **above 400**
costs **5 baht each**. Read the number of `units` used and print the total bill.

**Your program must**
- read `units` (a whole number, 0 or more)
- use `if` / `elif` / `else` to find which tier the usage reaches
- charge each tier at its own rate — only the units that fall inside that tier
- print `Total bill: X baht`

**Sample runs**
```
Units: 100              Units: 300              Units: 500
Total bill: 300 baht    Total bill: 1050 baht   Total bill: 1950 baht
```

*Hint:* write the total for each band first — up to 150 → `units*3`; 151–400 → `150*3 + (units-150)*4`;
above 400 → `150*3 + 250*4 + (units-400)*5`. Then use `if/elif/else` to pick the right one.

````python
# Write your program here
unit=int(input("Units : "))
if unit < 0:
  print ("Error")
elif unit<=150:
  totalbill=unit*3
  print (f"Totalbill: {totalbill} baht ")
elif 150 < unit <= 400:
  totalbill1=150*3 + (unit-150)*4
  print ((f"Totalbill: {totalbill1} baht "))
else:
  totalbill2=150*3 + 250*4 +(unit-400)*5
  print ((f"Totalbill: {totalbill2} baht "))

````
